```python
names = input("Enter first names separated by spaces: ").split()

count = 0
for name in names:
    count += name.lower().count('a')

print("Names:", names)
print("Occurrences of 'a':", count)
```

```python
color_list1 = input("Enter colors for list 1: ").split()
color_list2 = input("Enter colors for list 2: ").split()

result = [color for color in color_list1 if color not in color_list2]

print("Colors not in list 2:", result)
```

```python
n = int(input("Enter number of dictionary items: "))
my_dict = {}

for i in range(n):
    key = int(input("Enter key: "))
    value = input("Enter value: ")
    my_dict[key] = value

ascending = dict(sorted(my_dict.items()))
descending = dict(sorted(my_dict.items(), reverse=True))

print("Ascending order:", ascending)
print("Descending order:", descending)
```

```python
n1 = int(input("Enter number of items in dictionary 1: "))
dict1 = {}

for i in range(n1):
    key = input("Enter key: ")
    value = input("Enter value: ")
    dict1[key] = value

n2 = int(input("Enter number of items in dictionary 2: "))
dict2 = {}

for i in range(n2):
    key = input("Enter key: ")
    value = input("Enter value: ")
    dict2[key] = value

merged_dict = {**dict1, **dict2}

print("Dictionary 1:", dict1)
print("Dictionary 2:", dict2)
print("Merged dictionary:", merged_dict)
```
