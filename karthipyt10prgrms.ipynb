```python
s = input("Enter a string: ")

if s.startswith("Python"):
    print("The string starts with Python")
else:
    print("The string does not start with Python")
```

```python
s = input("Enter a string: ")

if s.endswith("ing"):
    print("The string ends with ing")
else:
    print("The string does not end with ing")
```

```python
s = input("Enter a string: ")
ch = input("Enter the character to find: ")

position = s.find(ch)
print("Position:", position)
```

```python
s = input("Enter a string: ")

if s.isalpha():
    print("The string contains only alphabets")
else:
    print("The string does not contain only alphabets")
```

```python
s = input("Enter a string: ")

if s.isdigit():
    print("The string contains only digits")
else:
    print("The string does not contain only digits")
```
